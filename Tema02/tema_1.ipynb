Extensie Python: Simulaţi 100.000 de aruncări şi estimaţi probabilităţile condiţionate. Comparaţi rezultatele cu valorile exacte.

```python
import numpy as np

rng = np.random.default_rng()
nr_aruncari = 100_000
aruncari = rng.integers(low=1, high=7, size=(nr_aruncari, 2))

cond_primul_par = aruncari[:, 0] % 2 == 0
cond_aldoilea_par = aruncari[:, 1] % 2 == 0
cond_sumaminim10 = aruncari.sum(axis=1) >= 10
cond_suma8 = aruncari.sum(axis=1) == 8

sansa_sumaminim10_primulpar = np.count_nonzero(cond_primul_par & cond_sumaminim10) / nr_aruncari * 100
sansa_primulpar = np.count_nonzero(cond_primul_par) / nr_aruncari * 100
print(f"1) P(S >= 10 | P1) = P(S >= 10, P1) / P(P1) = {sansa_sumaminim10_primulpar / sansa_primulpar :.5f}% (2/9)")

sansa_sumaminim10_celputin1par = np.count_nonzero((cond_primul_par | cond_aldoilea_par) & cond_sumaminim10) / nr_aruncari * 100
sansa_celputin1par = np.count_nonzero(cond_primul_par | cond_aldoilea_par) / nr_aruncari * 100
print(f"2) P(S >= 10 | P1 U P2) = P(S >= 10, P1 U P2) / P(P1 U P2) = {sansa_sumaminim10_celputin1par / sansa_celputin1par :.5f}% (5/27)")

sansa_suma8_ambelepare = np.count_nonzero(cond_suma8 & cond_primul_par & cond_aldoilea_par) / nr_aruncari * 100
sansa_suma8 = np.count_nonzero(cond_suma8) / nr_aruncari * 100
print(f"3) P(P1, P2 | S = 8) = P(P1, P2, S = 8) / P(S = 8) = {sansa_suma8_ambelepare / sansa_suma8 :.5f}% (3/5)")
```
