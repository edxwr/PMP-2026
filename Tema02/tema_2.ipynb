**Ex. 2 Filtrarea mesajelor spam**  
**Un sistem de e-mail primeşte 30% mesaje spam. Cuvântul „ofertă” apare în 80% dintre mesajele spam şi în 15% dintre
mesajele legitime.**
  * P(Spam) = 3/10
  * P(&not;Spam) = P(Legit) = 1 - 3/10 = 7/10
  * P("Oferta" | Spam) = 8/10
  * P("Oferta" | Legit) = 15/100

1. **Care este probabilitatea ca un mesaj care conţine cuvântul „ofertă” să fie spam?**
    * **P(Spam | "Oferta") = ?**
        * P("Oferta") =  
        P(Spam) * P("Oferta" | Spam) + P(Legit) * P("Oferta" | Legit) =  
        3/10 * 8/10 + 7/10 * 15/100 =  
        24/100 + 105/1000 =  
        240/1000 + 105/1000 =  
        345/1000 = 69/200
        * **P(Spam | "Oferta") =**  
        P(Spam, "Oferta") / P("Oferta") =  
        P(Spam) * P("Oferta" | Spam) / P("Oferta") =  
        3/10 * 8/10 / (69/200) =  
        24/100 * 200/69 =  
        8 * 2/23 **= 16/23**


2. **Presupunem că termenul „gratuit” apare în 60% dintre mesajele spam şi în 10% dintre mesajele legitime. Considerând independenţa condiţionată a apariţiei celor două cuvinte, calculaţi probabilitatea ca un mesaj care conţine ambele cuvinte să fie spam.**
