**Ex. 3 Paradoxul lui Simpson**  
O universitate analizează rata de promovare la două discipline, A şi B, pentru studenţii din două programe de studiu:  
|Program    |Disciplina A |Disciplina B|
|-----------|-------------|------------|
|Programul 1|  81 din 90  | 19 din 20  |
|Programul 2|  1 din 10   | 72 din 80  |

Valorile reprezintă numărul studenţilor promovaţi din totalul celor evaluaţi.

1. **Calculaţi rata de promovare la fiecare disciplină, separat pentru fiecare program.**
* P(A, Promovat | P1) = 81/90 = 9/10
* P(B, Promovat | P1) = 19/20
* P(A, Promovat | P2) = 1/10
* P(B, Promovat | P2) = 72/80 = 9/10

2. **Care disciplină are o rată de promovare mai mare în fiecare program?**
* P(B, Promovat | P1) > P(A, Promovat | P1)
    * Pentru programul 1, disciplina B are o rata de promovare mai mare
* P(B, Promovat | P2) > P(A, Promovat | P2)
    * Pentru programul 2, disciplina B are o rata de promovare mai mare

3. **Calculaţi rata globală de promovare pentru cele două discipline.**
* P(A, Promovat) = 82/100
* P(B, Promovat) = 91/100

4. **Explicaţi de ce ordinea disciplinelor se inversează atunci când agregăm datele.**
* In exemplul nostru nu s-au inversat datele, rata globala de promovare pentru Disciplina B e tot mai mare decat cea globala pentru A. Media globala pentru B este intre 90% si 95% iar pentru A este intre 10% si 90%.
* Astfel niciodata nu vom putea trage media globala a lui A sa fie mai mare decat media globala a lui B.
* Daca am fi avut:
    * P(A, Promovat | P1) = 30% si P(A, Promovat | P2) = 50%
    * P(B, Promovat | P1) = 40% si P(B, Promovat | P2) = 60%
* B ar fi avut procentaj mai bun per program dar daca am fi avut si o cantitate mare de studenti pentru P(A, Promovat | P2) si P(B, Promovat | P1) atunci la un moment dat media globala a lui A ar fi fost mai mare decat cea a lui B.

5. **Formulaţi fenomenul în termeni de probabilităţi condiţionate.**
* P(B, Promovat | P1) > P(A, Promovat | P1)
* P(B, Promovat | P2) > P(A, Promovat | P2)
* P(B, Promovat) < P(A, Promovat)

Extensie Python: Construiţi un exemplu propriu de paradox Simpson şi utilizaţi librăria pandas pentru a analiza
datele.

```python
import pandas as pd

date = {
    'Program': [1, 1, 2, 2],
    'Disciplina': ['A', 'B', 'A', 'B'],
    'Promovati': [30, 40_000, 50_000, 60], # am inmultit cu 1000 studentii din programele unde A are 50% si B 40%
    'Total': [100, 100_000, 100_000, 100] # ca sa trec media lui A peste cea a lui B
}
df = pd.DataFrame(date)

df['Procentaj'] = df['Promovati'] / df['Total']
print(df)
print('P1 => P(B) > P(A) si P2 => P(B) > P(A)')

global_df = df.groupby('Disciplina')[['Promovati', 'Total']].sum()
global_df['Procentaj'] = global_df['Promovati'] / global_df['Total']
print(global_df)
print('Cand agregam datele => P(A) > P(B)')
```
