**Ex.1 a)**

```python
import numpy as np


def simulare_ex1() -> bool:
    r = 3
    a = 4
    n = 2

    zar = np.random.randint(1, 7)
    if zar in [2, 3, 5]:
        # print("Am adaugat o bila neagra")
        n += 1
    elif zar == 6:
        # print("Am adaugat o bila rosie")
        r += 1
    else:
        # print("Am adaugat o bila albastra")
        a += 1

    bila_index = np.random.randint(0, r + a + n)
    if 0 <= bila_index < r:
        # print("Am extras o bila rosie")
        r -= 1
        return True
    elif r <= bila_index < r + a:
        # print("Am extras o bila albastra")
        a -= 1
    else:
        # print("Am extras o bila neagra")
        n -= 1

    return False

    # print("Situatia bilelor dupa extragere:")
    # print(f"{r} bile rosii")
    # print(f"{a} bile albastre")
    # print(f"{n} bile negre")
```

**Ex.1 b)**

```python
nr_incercari = 10000
nr_succese = sum(simulare_ex1() for _ in range(nr_incercari))
probabilitate_estimata = nr_succese / nr_incercari
print(f"Probabilitate estimata extragere bila rosie: {probabilitate_estimata*100}%")
```
