```python
import networkx as nx
import numpy as np
import pyAgrum as gum
import pyAgrum.lib.notebook as gnb

# 1. Crearea rețelei Markov
# Notă: În versiunile recente de pyAgrum se folosește gum.MRF(), iar în cele mai vechi gum.MarkovNet()
mn = gum.MRF() if hasattr(gum, "MRF") else gum.MarkovNet()

# 2. Adăugarea celor 5 variabile discrete binare (cu valorile 0 și 1)
for i in range(1, 6):
  mn.add(gum.LabelizedVariable(f"A{i}", f"A{i}", ["0", "1"]))

# 3. Definirea clicilor maximale ca factori în rețeaua Markov
maximal_cliques = [
    ["A1", "A2"],
    ["A1", "A3"],
    ["A3", "A4"],
    ["A2", "A4", "A5"],
]

for clique in maximal_cliques:
  mn.addFactor(clique)

# 4. Vizualizarea grafului în Jupyter Notebook
gnb.showMRF(mn)  # sau gnb.show(mn)

# 5. Determinarea automată a clicilor maximale folosind NetworkX pe baza muchiilor
edges = [
    ("A1", "A2"),
    ("A1", "A3"),
    ("A2", "A4"),
    ("A2", "A5"),
    ("A3", "A4"),
    ("A4", "A5"),
]
G = nx.Graph(edges)
print("Toate clicile:", list(nx.enumerate_all_cliques(G)))
print("Clicile maximale:", list(nx.find_cliques(G)))
```
